```python
from scipy.stats import chi2_contingency, kruskal
from ses.pipeline import cargar_entrenamiento, NUMERICAS, NOMINALES, BINARIAS

orden = ["Pobre", "Vulnerable", "Media", "Alta"]
y = y.map(dict(enumerate(orden)))

# ---------- Tabla nominales y binarias (chi-cuadrado) ----------
filas = []
for variable in NOMINALES + BINARIAS:
    conteo = pd.crosstab(X[variable], y)[orden]
    pct = pd.crosstab(X[variable], y, normalize="columns")[orden] * 100
    chi2, p, gl, esperados = chi2_contingency(conteo)

    celdas = conteo.astype(str) + " (" + pct.round(1).astype(str) + " %)"
    if variable in BINARIAS:
        celdas = celdas.loc[[1]]                      # solo la fila "sí"
    celdas.index = [f"{variable} = {c:g}" for c in celdas.index]

    celdas["Prueba"] = ""
    celdas["p"] = ""
    celdas["Pasa (p<0,25)"] = ""
    celdas.iloc[0, -3:] = ["Chi-cuadrado", f"{p:.3g}", "Sí" if p < 0.25 else "No"]
    filas.append(celdas)

tabla_cat = pd.concat(filas)
tabla_cat.index.name = "Variable"

# ---------- Tabla numéricas (Kruskal-Wallis) ----------
filas = []
for variable in NUMERICAS:
    grupos = [X.loc[y == clase, variable] for clase in orden]
    estadistico, p = kruskal(*grupos)

    fila = {"Variable": variable}
    for clase, g in zip(orden, grupos):
        fila[clase] = f"{g.median():.2f} ({g.quantile(0.25):.2f}–{g.quantile(0.75):.2f})"
    fila["Prueba"] = "Kruskal-Wallis"
    fila["p"] = f"{p:.3g}"
    fila["Pasa (p<0,25)"] = "Sí" if p < 0.25 else "No"
    filas.append(fila)

tabla_num = pd.DataFrame(filas).set_index("Variable")

tabla_cat.to_csv("reports/tables/tabla1a_categoricas.csv")
tabla_num.to_csv("reports/tables/tabla1b_numericas.csv")
```
