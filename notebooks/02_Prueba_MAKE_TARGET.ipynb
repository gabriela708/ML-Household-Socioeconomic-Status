```python
clase_porcentajes=d['ses'].value_counts()/19690
ses_porcentajes=clase_porcentajes.round(3).sort_index()*100
```

```python
import pandas as pd

distribucion_clases = pd.DataFrame(ses_porcentajes)
distribucion_clases = pd.concat([distribucion_clases, d['ses'].value_counts()], axis=1)
distribucion_clases = distribucion_clases.reset_index()
distribucion_clases.columns = ["ses", "distribucion(%)", "conteo(cabezeras)"]
distribucion_clases
```
